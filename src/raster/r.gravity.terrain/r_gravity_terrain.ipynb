# r.gravity.terrain

This notebook runs the example from r.gravity.terrain's manual page and visualizes the output.

## Setup

We will be using the NC SPM sample location.

```python
import os
import json
import subprocess
import sys

# Ask GRASS where its Python packages are.
sys.path.append(
    subprocess.check_output(["grass", "--config", "python_path"], text=True).strip()
)

# Import GRASS packages
import grass.script as gs
import grass.jupyter as gj

# Start GRASS Session
gj.init("~/data/grassdata/nc_basic_spm_grass7/user1")
```

## Create example from the tool's documentation

```python
r_gravity_terrain_input = "elevation"
r_gravity_terrain_output = "output"
gs.run_command(
    "r.gravity.terrain",
    input=r_gravity_terrain_input,
    output=r_gravity_terrain_output,
)
```

```python
r_gravity_terrain_map = gj.Map()
r_gravity_terrain_map.d_rast(map=r_gravity_terrain_output)
r_gravity_terrain_map.show()
```
